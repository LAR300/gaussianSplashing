```python
from PIL import Image
import cv2
import tifffile
```

```python
image1 = ""
image2 = ""
```

```python

# Function to get image size
def get_image_size(image_path):
    img = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)  # Read image
    if img is not None:
        height, width = img.shape[:2]
        print(f"Image: {image_path}")
        print(f"Width: {width} pixels")
        print(f"Height: {height} pixels")
    else:
        print(f"Failed to open {image_path}")


# Get sizes
get_image_size(image_path=image1)
```

```python
# Function to get TIFF image size
def get_tiff_size(image_path):
    with tifffile.TiffFile(image_path) as tif:
        image = tif.pages[0]  # Access the first page
        width, height = image.shape[1], image.shape[0]
        print(f"Image: {image_path}")
        print(f"Width: {width} pixels")
        print(f"Height: {height} pixels")
        print(f"Number of Pages: {len(tif.pages)}")
        
get_image_size(image_path=image2)
```
